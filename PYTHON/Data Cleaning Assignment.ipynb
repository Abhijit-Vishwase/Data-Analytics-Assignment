Q9. Time-Based Missing Handling
Scenario: 
Admission dates should follow a logical sequence.
Task:
Sort data by Admission_Date
Apply forward fill or backward fill where appropriate
Justify your choice

```python
df['Admission_Date'] = pd.to_datetime(df['Admission_Date'])
df=df.sort_values(by='Admission_Date')
df
```

```python
df['Treatment_Cost'] = df['Treatment_Cost'].fillna(method='ffill').fillna(method='bfill')
```

```python
df['Treatment_Cost']
```

```python
This is a two-stage imputation strategy applied after sorting by Admission_Date. The justification is grounded in temporal consistency and completeness.

Primary logic: forward fill (ffill)

Assumes that the most recent known value is the best estimate for a missing value at time t
Respects causality: only uses past information
Appropriate for time-sequenced operational data (like admissions), where values tend to evolve gradually rather than jump arbitrarily

Why not only forward fill?

If the dataset starts with missing values, there is no prior observation to propagate
Those leading NaNs would remain unfilled

Secondary logic: backward fill (bfill)

Applied only to remaining nulls after ffill
Fills initial gaps using the nearest future value
Ensures no missing values remain, which is often required for modeling pipelines
```
