Q2 Sales Filtering Analysis
Filter the dataset to display:
Orders where Sales > 3500
Orders where Region = "East" and Sales > 2500
Explain what insights you observe.

```python
df[df['Sales']>3500]
```

Insights from Sales > 3500
High-value orders are mostly concentrated in Electronics products such as Phones, Laptops, and Monitors.
These orders contribute significantly to overall revenue, indicating that a small number of transactions drive a large portion of sales (high-value sales concentration).
However, high sales do not always guarantee high profit—some transactions still show reduced or negative profit due to discounts.
This suggests that pricing and discount strategy directly impacts profitability, even in premium orders.

```python
df[(df['Region'] =='East') & (df['Sales']>2500)]
```

Insights from Region = "East" and Sales > 2500
The East region shows strong sales performance, with many orders crossing the 2500 threshold.
Cities like Ahmedabad, Kolkata, Chennai, and Delhi appear frequently, indicating key revenue-generating markets in this region.
The majority of these orders belong to the Electronics category, confirming it as the dominant revenue driver.
There are instances of negative or low profit despite high sales, mainly due to higher discounts (15–20%), highlighting margin issues.
Discount sensitivity is evident—higher discounts tend to reduce profitability significantly.
