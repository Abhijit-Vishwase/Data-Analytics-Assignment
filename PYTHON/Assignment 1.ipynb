Question 2 : Explain the difference between List and Tuple in Python.

Solution : 
Difference between List and Tuple

A list and a tuple are both data structures in Python used to store multiple items, but they differ mainly in mutability and usage.
A list is mutable, which means its elements can be changed, added, or removed after creation. Lists are written using square brackets [ ]. They are used when the data needs to be modified frequently.
A tuple, on the other hand, is immutable, meaning once it is created, its elements cannot be changed. Tuples are written using parentheses ( ). They are generally used when the data should remain constant and secure.
In terms of performance, tuples are slightly faster than lists because they are immutable. Lists have more built-in methods compared to tuples, which makes them more flexible.

Question 3 : What is a function in Python? Why are functions useful?

Solution : 
A function in Python is a block of organized, reusable code that is used to perform a specific task. 
Functions are defined using the def keyword and can be called whenever needed in a program.
Functions are useful because they help in reducing code repetition, making programs shorter and more efficient.
They improve readability and make the code easier to understand. Functions also help in organizing large programs into smaller, manageable parts, which makes debugging and maintenance easier.

Question 8 : Write a Python program to remove duplicate values from a list using a set.
