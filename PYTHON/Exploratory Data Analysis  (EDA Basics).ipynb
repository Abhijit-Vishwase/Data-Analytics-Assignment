## Assignment 5 : Exploratory Data Analysis (EDA Basics)

```python
import pandas as pd
df = pd.read_csv("C:/Users/abhij/Downloads/House Price India.csv")

print(df.info())
print(df.describe())
```

```python
'''Q1. What is the average, median, and standard deviation of house prices?'''
```

```python
print("Mean:", df['Price'].mean())
print("Median:", df['Price'].median())
print("Std:", df['Price'].std())
```

```python
'''Q2. Which number of bedrooms is most common?'''
```

```python
print(df['number of bedrooms'].mode())
```

```python
'''Q3. Check if the price data is skewed'''
```

```python
print(df['Price'].skew())
```

```python
'''Q4. What is the average price for each number of bedrooms?'''
```

```python
print(df.groupby('number of bedrooms')['Price'].mean())
```

```python
'''Q5.  What is the relationship between living area and price? (Use : Area without Basement)'''
```

```python
print(df[['Area of the house(excluding basement)','Price']].corr())
```

```python
'''Q6. Identify any anomalies where houses have high prices but low areas.'''
```

```python
anomalies = df[(df['Price'] > df['Price'].quantile(0.9)) & 
               (df['Area of the house(excluding basement)'] < df['Area of the house(excluding basement)'].quantile(0.3))]

print(anomalies)
```

```python
print(df.groupby(['number of floors','waterfront present'])['Price'].mean())
```

```python
print("Min:", df['Price'].min())
print("Max:", df['Price'].max())
```
